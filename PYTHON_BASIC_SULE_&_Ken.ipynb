# PYTHON OPERATOR

An operator is a special symbol that determines the kind of action that should be performed on a value or varialbe in a python program.

In python,, operators are grouped into the following categories:


(1) **Assignment Operator**: The assignment is the single qual sign(=). This operator is used to assign a value to a variable in python program.

(2) **Arithmetic Operator**: These operator are used to perform mathematical calculations, and they are:
       (1)  +: which is used to perform addition of two or more numbers
       (2)  *: which is used to perform multiplication of two or more numbers

       (3) /:  which is used to perform divide two or more numbers

       (4) -:  which is used to perform substract two or more numbers from a given number

       (5) Modulus (%): This operator returns the remainder after dividing two numbers
    
  (3) **Membership:** This operator is used to check if a given number or string exist within a given variable.

  (4) **Logical:** Python logical operators are used to combine conditional statements and evaluate Boolean expressions e.g.and, or, not    

  (5) **Relational/Comparison**: These operators are used to compare two or more values or variable, e.g. <, >, <=, >=, ==, !=

  (6) **Bitwise:** Bitwise operators in Python are used to perform operations directly on the individual bits (binary digits) of integer data types. These operators treat numbers as a sequence of ones and zeros, allowing for low-level data manipulation, which can be useful for tasks like setting flags, data compression, or optimization. Examples are &, |, ~, ^ etc.

```python


stored_email =  "michaelanietiejack@gmail.com"
stored_password = "12345678"

max_attempt =3
attempts_left = max_attempt

while attempts_left > 0:
  email = input("\nEmail: ").strip() # the strip() is used to trim off while spaces within an input
  password = input("\npassword: ").strip() # the strip() is used to trim off while spaces within an input

  # 1. validation 1
  if stored_email =="" and stored_password == "":
    print("Error: Email and Password are required.")
    continue
   # 2. validation 2
  elif email == "":
        print("Error: Email is required.")
        continue

  elif password == "":
        print("Error: Password is required.")
        continue

  if "@" not in email or "." not in email:
        print("Error: Invalid email format (e.g., student@example.com).")
        continue

  if len(password) < 8:
        print("Error: Password must be at least 8 characters.")
        continue

  if email == stored_email and password == stored_password:
        print("Login successful ✅ Welcome!")
        break

  else:
        attempts_left -= 1
        print(f"Invalid login details ❌ Attempts left: {attempts_left}")

  if attempts_left == 0:
            print("Account locked 🔒 Too many failed attempts.")
```

Write a python function to find the factorial of any given given number

```python
def factorial(n):
    """
    Calculates factorial using recursion.
    """
    if n < 0:
        return "Factorial is not defined for negative numbers"
    if n == 0 or n == 1:
        return 1

    return n * factorial(n - 1)
```

```python
def factorial_loop(n):
    """
    Calculates the factorial of a non-negative integer n.
    """
    if n < 0:
        return "Factorial is not defined for negative numbers"
    result = 1


    for i in range(1, n + 1):
        result *= i

    return result
```

```python
# calling the function to find thefactorial of a number
factorial_loop(5)
```

**Default Arguments**

A default argument is a parameter that already has a value in the function definition.

If the caller does not provide that argument, Python uses the default value.

```python
def power(base=5, exponent=2):
    return base ** exponent

print(power(5,3))      # uses exponent=2

```
